```python
df.drop(columns=['show_id'], inplace=True)
```

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')
```

```python
df['year_added'] = df['date_added'].dt.year
```

```python
df['duration_value'] = df['duration'].str.extract(r'(\d+)').astype(float)
```

```python
df.rename(columns={
    'listed_in': 'genre',
    'release_year': 'year'
}, inplace=True)
```

```python
print(df.head())
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
print(df.isnull().sum().sum())
```

```python
df['year'] = df['year'].fillna(df['year'].mean())
```

```python
df['director'] = df['director'].fillna('Unknown')

df['cast'] = df['cast'].fillna('Unknown')

df['country'] = df['country'].fillna('Unknown')
```

```python
print(df.isnull().sum())
```

```python
print(df['type'].value_counts())
```

```python
country_count = df['country'].value_counts().head(10)

print(country_count)
```

```python
genre_count = df['genre'].str.split(', ').explode().value_counts().head(10)

print(genre_count)
```

```python
yearly_content = df.groupby('year_added')['title'].count()

print(yearly_content)
```

```python
print(df['rating'].value_counts().head(10))
```

```python
year_type = df.groupby(['year_added', 'type'])['title'].count()

print(year_type)
```

```python
import matplotlib.pyplot as plt
```

```python
content_trend = df.groupby('year_added')['title'].count().reset_index()

plt.figure(figsize=(12,6))

plt.plot(content_trend['year_added'],
         content_trend['title'],
         marker='o')

plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.title('Netflix Content Added Each Year')

plt.grid()
plt.show()
```

```python
type_count = df['type'].value_counts()

plt.figure(figsize=(7,5))

plt.bar(type_count.index, type_count.values)

plt.xlabel('Content Type')
plt.ylabel('Number of Titles')
plt.title('Movies vs TV Shows')

plt.show()
```

```python
country_count = df['country'].str.split(', ').explode().value_counts().head(10)

plt.figure(figsize=(12,6))

plt.bar(country_count.index, country_count.values)

plt.xlabel('Country')
plt.ylabel('Number of Titles')
plt.title('Top 10 Countries on Netflix')

plt.xticks(rotation=45)
plt.show()
```

```python
genre_count = df['genre'].str.split(', ').explode().value_counts().head(10)

plt.figure(figsize=(12,6))

plt.bar(genre_count.index, genre_count.values)

plt.xlabel('Genre')
plt.ylabel('Number of Titles')
plt.title('Top 10 Netflix Genres')

plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(6,6))

plt.pie(
    type_count.values,
    labels=type_count.index,
    autopct='%1.1f%%'
)

plt.title('Netflix Content Distribution')
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(df['year'].dropna(), bins=20)

plt.xlabel('Release Year')
plt.ylabel('Number of Titles')
plt.title('Netflix Release Year Distribution')

plt.show()
```
